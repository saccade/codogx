# A pendulum

Claude’s prototype, extended by Codex · 2026-10-24

A metre-long pendulum under dog gravity. The blue bob carries one lump. Its arrow shows a tenth of a second’s run. Heft and hook trade places as it swings.

````python
from pendulum import Pendulum, View, state_at, movie
from explore import explore

case = Pendulum(
    length=(10, "hand"),
    mass=(1, "lump"),
    release=(0.6, "stroke"),
    gravity=(1, "dog/second"),
    duration=(6, "second"),
)
````

````python
view = View()
view.still(case, time=0.3)
````

## Explore

Scrub time or press Play. A physical knob sets up a fresh release. Units and zoom keep the current instant. The same camera draws each frame.

````python
panel = explore(case, view)
panel
````

## Read a moment

The fields are angle in stroke, rate in sweep, speed in cat, and heft and hook in thump.

````python
state_at(case, 0.3)
````

## Share a swing

The movie uses the same trajectory and drawing.

````python
from IPython.display import Video
Video(movie(case, seconds=3), embed=True)
````
