# A charge's velocity, bent by flux

Written by Claude on Codex's interface · 2026-10-24

One coulomb on one lump, set off forward at 3 cat through 5 weber stroke a sheet coming up out of the page. Flux per sheet is an axis arrow, so it carries a stroke, and the cross with the charge's run takes it away again. Each gray mark is one weber. The teal arrow is a tenth of a second's run; the red one is the run the push adds in that tenth. The push stays square to the run, so the speed and the punch never change: the flux only bends.

Pillow rendering and GIF export revised by Codex · 2026-10-24.

```python
from flux import ChargeInFlux, View, state_at, explore, movie

case = ChargeInFlux(
    charge=(1, "coulomb"),
    mass=(1, "lump"),
    speed=(3, "cat"),
    field=(5, "weber*stroke/sheet"),
    duration=(6, "second"),
)
```

```python
view = View()
view.still(case, time=0.4)
```

## Explore

Scrub time or press Play. A knob sets the charge off again from the top. Turn the field below zero and the marks become crosses: the flux goes down into the page, and the bend turns forward around. A negative charge does the same.

```python
panel = explore(case, view)
panel
```

## Read a moment

Time in seconds, place in meters from the center, velocity in cat, push in push, punch in thump, and the bend in strokes a second, forward around.

```python
state_at(case, 0.4)
```
